```python
from sklearn.model_selection import train_test_split

# ============================================================
# 1. Conversion de la colonne « abonnement »
# ============================================================
# On travaille sur une copie pour ne pas modifier df par effet de bord
# (et éviter qu'une ré-exécution de la cellule ne transforme tout en NaN).
df = df.copy()

# Nettoyage préalable : espaces et casse (" premium " -> "Premium")
df["abonnement"] = df["abonnement"].astype(str).str.strip().str.capitalize()

mapping_abonnement = {"Basique": 0, "Standard": 1, "Premium": 2}
df["abonnement"] = df["abonnement"].map(mapping_abonnement)

# Vérification : toute valeur non reconnue devient NaN -> on le signale
valeurs_inconnues = df["abonnement"].isna().sum()
if valeurs_inconnues > 0:
    raise ValueError(
        f"{valeurs_inconnues} valeur(s) d'abonnement non reconnue(s). "
        "Vérifiez l'orthographe ou les valeurs manquantes."
    )

# ============================================================
# 2. Définition des variables explicatives (features)
# ============================================================
features = ["age", "salaire", "abonnement", "anciennete", "consommation"]
X = df[features]

# ============================================================
# 3. Définition de la variable cible (target)
# ============================================================
y = df["churn"]

# ============================================================
# 4. Séparation Train / Test
# ============================================================
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,      # 20 % des données réservées au test
    random_state=42,    # reproductibilité
    stratify=y          # conserve la proportion de churn dans train et test
)

print(f"Train : {X_train.shape} | Test : {X_test.shape}")
print("Proportion de churn (train) :", round(y_train.mean(), 3))
print("Proportion de churn (test)  :", round(y_test.mean(), 3))
```
